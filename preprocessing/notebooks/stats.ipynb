```python
import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
LOGS_DIR = "/gpfs/commons/home/vmazeeva/gruyere-expr/preprocess/logs"

TRAIN_COHORTS = ["GTEX", "AnswerALS", "Mayo", "NYGC", "MSBB", "ROSMAP"]
TEST_COHORT = ["ROSMAP_DLPFC"]
```

#### SNP Filtering

```python
logs = os.path.join(LOGS_DIR, "subset")

cohort_snps_stats = dict()

for cohort in TRAIN_COHORTS + TEST_COHORT:
    snp_stat_section = False
    total, snp, indel = 0, 0, 0
    log_file = os.path.join(logs, f"subset_genotype_{cohort}.out")
    with open(log_file, "r") as f:
        for line in f.readlines():
            if line.startswith("SNP Filtering Statistics"):
                snp_stat_section = True
                f.readline() # skip ===== line
                continue
            if snp_stat_section:
                if line.startswith("="): continue
                if line.strip().startswith("Total"):
                    total = line.split(":")[-1].strip()
                elif line.strip().startswith("SNPs"): 
                    snp = line.split(":")[-1].strip()
                elif line.strip().startswith("Indels"):
                    indel = line.split(":")[-1].strip()
                else:
                    cohort_snps_stats[cohort] = (total, snp, float(int(snp)/int(total)))
                    break

```
