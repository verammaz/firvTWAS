```python
import os
GENOTYPE_DIR = '/gpfs/commons/groups/knowles_lab/vmazeeva/BigBrain/Processed/genotypes/'

genes = [] 

for chrom in range(1,23):
    g_genes = [f"chr{chrom}/{f.split('_')[0]}\n" for f in os.listdir(os.path.join(GENOTYPE_DIR, f"chr{chrom}")) 
                if f.endswith("_genotypes.tsv.gz")]
    genes.extend(g_genes)
```

```python
with open('gene_list_all.txt', 'w') as f:
    f.writelines(genes)
```
